```python
#Mutability in Python refers to the ability of an object to be changed 
# after it has been created. 

list1 = [1, 2, 3]

list2 = list1

list2.append(4)

print(list1)  # Output: [1, 2, 3, 4]
print(list2)  # Output: [1, 2, 3, 4]


# this has to do with the fact that lists are mutable objects in Python. 
# When you assign list1 to list2, 
# both variables point to the same list object in memory. 
# Therefore, when you modify list2 by appending an element, 
# it also affects list1 since they reference the same object.

# THIS ALSO APPLIES TO DICTIONARIES AND SETS, 
# -----WHICH ARE MUTABLE OBJECTS IN PYTHON.------
```
